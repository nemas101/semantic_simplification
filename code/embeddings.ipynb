```python
import os

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import spacy
from sklearn.manifold import TSNE
```

```python
os.makedirs("../results/test/", exist_ok=True)
```

```python
tsne = TSNE(n_components=2, perplexity=2)
```

```python
nlp = spacy.load(name="en_core_web_lg")
```

```python
df = pd.read_json("../results/results.json")
```

```python
df = pd.read_json("../results/results.json")
parsed_text = nlp(df.iloc[0].original_text)
filtered_text = [token for token in parsed_text if not token.is_stop]
filtered_text = [token for token in filtered_text if not token.is_punct]
filtered_text = [token for token in filtered_text if not token.is_space]
vectors = [token.vector for token in filtered_text]
embeddings = np.array(vectors)
embeddings2d = tsne.fit_transform(embeddings)
plt.figure(figsize=(10, 7), dpi=1000)
# TODO display different POS tags differently
plt.scatter(embeddings2d[:, 0], embeddings2d[:, 1], marker='o')
for i, word in enumerate(filtered_text):
    plt.text(embeddings2d[i, 0], embeddings2d[i, 1],
            word, fontsize=10, ha='left', va='bottom')
plt.xlabel('t-SNE Dimension 1')
plt.ylabel('t-SNE Dimension 2')
plt.title('Word Embedding Graph (t-SNE with SpaCy)')
plt.grid(True)
plt.savefig(f'../results/embeddings.png')
plt.show()
    
```
